# 📱 Análisis estadístico de planes móviles

Análisis exploratorio y pruebas estadísticas para comparar ingresos entre planes
móviles.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

## 📌 Carga de datos

```python
df = pd.read_csv("data.csv")  # ajusta el nombre real del archivo
df.head()
```

## 📊 Análisis exploratorio

```python
df.groupby("plan")["revenue"].mean()
```

## 🧪 Prueba de hipótesis

```python
plan_a = df[df["plan"] == "A"]["revenue"]
plan_b = df[df["plan"] == "B"]["revenue"]

stat, p_value = stats.ttest_ind(plan_a, plan_b, equal_var=False)
p_value
```

## ✅ Conclusión

El análisis mostró diferencias estadísticamente significativas en los ingresos
generados por los planes móviles.
